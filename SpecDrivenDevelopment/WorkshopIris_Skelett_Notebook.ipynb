## 1 · Daten laden

Wir nutzen den klassischen **Iris-Datensatz**: 150 Schwertlilien, drei Arten, vier gemessene Merkmale (Länge und Breite von Kelch- und Blütenblatt).

Diese Zelle ist bewusst vorgegeben, damit **alle mit identischen Daten** starten – unabhängig von der Umgebung.

```python
# --- Daten laden: einfach ausführen, nichts ändern ---
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
df = iris.frame                      # DataFrame mit Merkmalen + Zielspalte 'target'
df["art"] = df["target"].map(dict(enumerate(iris.target_names)))

print(f"Zeilen: {len(df)}, Spalten: {list(df.columns)}")
df.head()
```
